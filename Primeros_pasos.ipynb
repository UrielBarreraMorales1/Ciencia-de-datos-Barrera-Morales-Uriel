```python
def clasificar_edad(edad):
    """Recibe una edad y regresa la categoria correspondiente."""
    if edad < 12:
        return "Niño"
    elif edad < 18:
        return "Adolescente"
    elif edad < 65:
        return "Adulto"
    else:
        return "Adulto Mayor"



for edad in [5,15,40,80]:
    print (edad,"->", clasificar_edad(edad))
```

```python
pasajeros = []

def agregar_pasajero():
    nombre = input("Nombre del pasajero: ")
    edad = int(input("Edad: "))
    clase = int(input("Clase: "))
    sobrevivio = input("¿Sobrevivió? (Si/No): ").lower() == "si"

    pasajero = {
        "nombre": nombre,
        "edad": edad,
        "clase": clase,
        "sobrevivio": sobrevivio
    }

    pasajeros.append(pasajero)
    print("Registro correcto.\n")


while len(pasajeros) < 4:
    print("¿Deseas agregar a un pasajero? (1=Sí / 2=No)")
    
    try:
        opcion = int(input("> "))

        if opcion == 1:
            agregar_pasajero()

        elif opcion == 2:
            print("Finalizando la captura...")
            break

        else:
            print("Ingresa solamente 1 o 2.\n")

    except ValueError:
        print("Error: Ingresa un número válido.\n")


print("\nLISTA FINAL DE PASAJEROS")
print(pasajeros)
```
