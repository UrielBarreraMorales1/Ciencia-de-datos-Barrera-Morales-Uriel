```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv('titanic.csv')
print("carga de datos exitoso")
```

```python
df.head()
```

```python
df.info() #tipos de datos que tenemos
```

```python
df.describe() #estadisticas basicas (media,desviacion estandar, min, max etc)
```

```python
df.describe(include='all')
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())#rellenar la comulna
```

```python
df.head(10)
```

```python
df.dropna(subset=['Embarked'],inplace=True) #eliminar filas
```

```python
df.drop('Cabin',axis=1, inplace=True) #Elminar columnas
```

```python
df['Pclass'] =df['Pclass'].astype('category') #convertit
```

```python
df.info()
```

```python
pip install seaborn
```

```python
import seaborn as sns
sns.set(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"]=(10,6)
```

```python
sns.countplot(x="Survived", data=df)
plt.title("Distribucion de sobrevivientes")
plt.show()
```

```python
cat_vars=["Sex","Pclass","Embarked"]
```

```python
for col in cat_vars:
    sns.countplot(x=col, data=df)
    plt.title("Distribucion de {col}")
    plt.show()
```

```python
num_vars=["Age", "Fare"]
```

```python
for col in num_vars:
    sns.histplot(df[col].dropna(), kde=True, bins=30)
    plt.title("Distribucion de {col}")
    plt.show()
```

```python
#Superviviencia por sexo
sns.barplot(x="Sex", y="Survived", data=df)
plt.title("Sobrevivientes por sexo")
plt.show()
```

```python
#Sobrevivencia por clase
sns.barplot(x="Pclass", y="Survived", data=df)
plt.title("Sobrevivientes segun clase")
plt.show()
```

```python
#Supervivencia por edad
sns.barplot(x="Survived", y="Age", data=df)
plt.title("Sobrevivientes segun edad")
plt.show()
```

```python
#Mapa de correlaciones
corr=df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Mapa de correlaciones")
plt.show()
```
